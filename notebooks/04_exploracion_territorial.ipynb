# Etapa 4 · Exploración territorial
Consulta rankings, semáforo y predicciones de Stage-II. Las zonas `sin_historico` no se interpretan como riesgo confiable.

```python
import pandas as pd
import matplotlib.pyplot as plt
scope = 'municipality'
data = pd.read_csv(f'../outputs/integrated/ccpc_{scope}_integrated.csv', low_memory=False)
data.head()
```

```python
data['risk_level'].value_counts(dropna=False).reindex(['bajo','medio','alto','sin_historico'], fill_value=0).plot.bar(title='Distribución del riesgo')
plt.xlabel('Nivel de riesgo'); plt.ylabel('Número de zonas'); plt.show()
```
